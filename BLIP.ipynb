```python
import torch
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import requests

# -------------------------------
# Device
# -------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
# -------------------------------
# Load BLIP model & processor
# -------------------------------
model_name = "Salesforce/blip-image-captioning-base"  # or use '-large' for bigger model

processor = BlipProcessor.from_pretrained(model_name, use_fast=True)
model = BlipForConditionalGeneration.from_pretrained(model_name).to(device)
```

```python

model.eval()

# -------------------------------
# Load an example image
# -------------------------------
url = "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/transformers/tasks/image_captioning.png"
# image = Image.open(requests.get(url, stream=True).raw).convert("RGB")
image = Image.open('dataset/flickr30k_images/train/134206.jpg').convert("RGB")

# -------------------------------
# Preprocess & generate caption
# -------------------------------
inputs = processor(images=image, return_tensors="pt").to(device)

# Generate caption ids
with torch.no_grad():
    output_ids = model.generate(**inputs)

# Decode caption
caption = processor.decode(output_ids[0], skip_special_tokens=True)
print("Caption:", caption)
```
