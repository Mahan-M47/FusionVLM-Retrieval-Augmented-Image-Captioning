## Import Libraries and Custom Configs

```python
import pandas as pd
import shutil
import random
import os

from pathlib import Path
```

```python
from Modules.config import (IMAGE_ROOT,
                            TRAIN_IMAGE_DIR,
                            TEST_IMAGE_DIR,
                            CAPTIONS_PATH,
                            TRAIN_CAPTIONS_PATH,
                            TEST_CAPTIONS_PATH,
                            N_TEST_IMAGES,
                            SEED)

random.seed(SEED)
```

## Load Data and Perform Train/Test Split

```python
# Load captions csv
df = pd.read_csv(Path(CAPTIONS_PATH))
df.head()
```

```python
# All unique image names (e.g. "1000092795.jpg")
all_images = df["image_name"].unique().tolist()

if len(all_images) < N_TEST_IMAGES:
    raise ValueError("Not enough images to create test split.")

test_images = set(random.sample(all_images, N_TEST_IMAGES))
train_images = set(all_images) - test_images
```

```python
# Move image files
def move_images(image_names, target_dir):
    for img_name in image_names:
        src_path = os.path.join(IMAGE_ROOT, img_name)
        dst_path = os.path.join(target_dir, img_name)

        if not os.path.exists(src_path):
            continue

        shutil.move(src_path, dst_path)

move_images(test_images, TEST_IMAGE_DIR)
move_images(train_images, TRAIN_IMAGE_DIR)
```

```python
# Split captions CSV
df_test = df[df["image_name"].isin(test_images)]
df_train = df[df["image_name"].isin(train_images)]

df_test.to_csv(TEST_CAPTIONS_PATH, index=False)
df_train.to_csv(TRAIN_CAPTIONS_PATH, index=False)
```

```python
# Summary
print(f"Test images:  {len(test_images)}")
print(f"Train images: {len(train_images)}")
print(f"Saved: {TEST_CAPTIONS_PATH}, {TRAIN_CAPTIONS_PATH}")
print("Image files moved successfully.")
```
