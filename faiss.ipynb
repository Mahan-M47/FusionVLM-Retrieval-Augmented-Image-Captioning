```python
import faiss
import numpy as np

# https://medium.com/@devbytes/similarity-search-with-faiss-a-practical-guide-to-efficient-indexing-and-retrieval-e99dd0e55e8c
```

```python
# -----------------------------
# 1. Sample data
# -----------------------------
embedding_dim = 4  # small dimension for demo
vectors = np.array([
    [0.1, 0.2, 0.3, 0.4],
    [0.9, 0.8, 0.7, 0.6],
    [0.5, 0.5, 0.5, 0.5]
], dtype=np.float32)

ids = ["vec1", "vec2", "vec3"]   # metadata linked to vectors
documents = ["doc1", "doc2", "doc3"]
```

```python
# -----------------------------
# 2. Create FAISS index
# -----------------------------
index = faiss.IndexFlatIP(embedding_dim)  # L2 distance
index.add(vectors)  # insert vectors
print("Initial number of vectors:", index.ntotal)
```

```python
# -----------------------------
# 3. Search vectors
# -----------------------------
query = np.array([[0.1, 0.2, 0.3, 0.4]], dtype=np.float32)
D, I = index.search(query, k=1)
print("Distances:", D)
print("Indices:", I)
for idx in I[0]:
    print("Matched vector ID:", ids[idx], "Doc:", documents[idx])
```

```python

# -----------------------------
# 4. Add more vectors
# -----------------------------
new_vectors = np.array([[0.2, 0.1, 0.0, 0.3]], dtype=np.float32)
index.add(new_vectors)
ids.append("vec4")
documents.append("doc4")
print("Number of vectors after add:", index.ntotal)
```
